```python
import pandas as pd
```

```python
auto = pd.read_json("../ex02/auto.json")
auto = auto.set_index("CarNumber")
auto.head()
```

```python
fines_over_2100 = auto.loc[auto["Fines"]>2100]
fines_over_2100.head()
```

```python
fines_over_2100_and_refunds_2=auto.loc[(auto["Fines"]>2100) & (auto["Refund"]==2)]
fines_over_2100_and_refunds_2.head()
```

```python
models = ["Focus", "Corolla"]

focus_or_corolla = auto.loc[
    auto["Model"].isin(models)
]

focus_or_corolla.head()
```

```python
numbers = ['Y7689C197RUS', '92928M178RUS', '7788KT197RUS', 'H115YO163RUS', 'X758HY197RUS']
numberss = auto.loc[auto.index.isin(numbers)]
numberss.head()
```

```python
median_fines_by_make = (
    auto
    .groupby("Make")["Fines"]
    .median()
)
median_fines_by_make
```

```python
median_fines_by_make_model = (
    auto
    .groupby(["Make", "Model"])["Fines"]
    .median()
)

median_fines_by_make_model
```

```python
fines_count_by_make_model = (
    auto
    .groupby(["Make", "Model"])["Fines"]
    .count()
)

fines_count_by_make_model
```

```python
fines_min_max = (
    auto
    .groupby(["Make", "Model"])["Fines"]
    .agg(["min", "max"])
)

fines_min_max
```

```python
fines_std_by_make_model = (
    auto
    .groupby(["Make", "Model"])["Fines"]
    .std()
)

fines_std_by_make_model
```

```python
fines_count_by_car = (
    auto
    .groupby(level="CarNumber")["Fines"]
    .count()
    .sort_values(ascending=False)
)

fines_count_by_car.head()
```

```python
top_car_by_count = fines_count_by_car.index[0]

top_car_by_count
```

```python
top_car_rows = auto.loc[[top_car_by_count]]

top_car_rows
```

```python
fines_sum_by_car = (
    auto
    .groupby(level="CarNumber")["Fines"]
    .sum()
    .sort_values(ascending=False)
)

fines_sum_by_car.head()
```

```python
top_car_by_sum = fines_sum_by_car.index[0]

top_car_by_sum
```

```python
top_sum_car_rows = auto.loc[[top_car_by_sum]]

top_sum_car_rows
```

```python
models_per_car = (
    auto
    .groupby(level="CarNumber")
    .agg(ModelCount=("Model", "nunique"))
)

models_per_car
```

```python
cars_with_multiple_models = models_per_car.loc[
    models_per_car["ModelCount"] > 1
]

cars_with_multiple_models
```

```python
auto.loc[
    auto.index.isin(cars_with_multiple_models.index),
    ["Make", "Model"]
]
```
