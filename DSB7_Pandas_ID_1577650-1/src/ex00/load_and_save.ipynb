## Импорт Pandas

```python
import pandas as pd
```

## Загрузка, переименование индекса и сохранение

```python
df = pd.read_csv(
    "../data/feed-views.log", 
    sep="\t", skiprows=[2, 3],
    skipfooter=2,
    engine="python",
    names=["datetime","user"],
    index_col="datetime"
)
df.index.name = "date_time"
df.to_csv(
    "../data/feed-views-semicolon.log",
    sep=";"
    )
```
