```python
import pandas as pd
```

```python
csv = pd.read_csv(
    "../../datasets/auto (1).csv",
    index_col="ID"
)
```

```python
csv.head()
```

```python
csv.count()
```

```python
csv = csv.drop_duplicates(
    subset=["CarNumber", "Make_n_model",  "Fines"],
    keep="last"
)
```

```python
csv.count()
```

```python
csv.isna().sum()
```

```python
csv=csv.dropna(
    axis=1,
    thresh=len(csv)-500
)
```

```python
csv.isna().sum()
```

```python
csv["Refund"] = csv["Refund"].ffill()
```

```python
csv["Refund"].isna().sum()
```

```python
fines_mean = csv["Fines"].mean()
csv["Fines"]=csv["Fines"].fillna(fines_mean)
```

```python
csv.isna().sum()
```

```python
parts = csv["Make_n_model"].apply(
    lambda text: text.split(maxsplit=1)
)

csv["Make"] = parts.apply(
    lambda parts: parts[0]
)

csv["Model"] = parts.apply(
    lambda parts: parts[1] if len(parts) > 1 else pd.NA
)

csv[["Make_n_model", "Make", "Model"]].head()
```

```python
csv = csv.drop(columns=["Make_n_model"])
```

```python
csv.to_json(
    "auto.json",
    orient="records",
    indent=2
)
```
