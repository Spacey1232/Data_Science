## Импорт Pandas

```python
import pandas as pd
```

## Чтение CSV

```python
csv = pd.read_csv(
    "../data/auto.csv",
    index_col="ID"
)
```

## Число исходных наблюдений

```python
csv.head()
```

## Удаление дубликатов

```python
csv.count()
```

## Число уникальных наблюдений

```python
csv = csv.drop_duplicates(
    subset=["CarNumber", "Make_n_model",  "Fines"],
    keep="last"
)
```

## Подсчёт пропусков

```python
csv.count()
```

## Удаление столбцов с пропусками

```python
csv.isna().sum()
```

## Проверка пропусков

```python
csv=csv.dropna(
    axis=1,
    thresh=len(csv)-500
)
```

## Заполнение Refund

```python
csv.isna().sum()
```

## Проверка Refund

```python
csv["Refund"] = csv["Refund"].ffill()
```

## Среднее значение Fines

```python
csv["Refund"].isna().sum()
```

## Заполнение Fines

```python
fines_mean = csv["Fines"].mean()
csv["Fines"]=csv["Fines"].fillna(fines_mean)
```

## Проверка заполнения

```python
csv.isna().sum()
```

## Разделение марки и модели

```python
parts = csv["Make_n_model"].apply(
    lambda text: text.split(maxsplit=1)
)

csv["Make"] = parts.apply(
    lambda parts: parts[0]
)

csv["Model"] = parts.apply(
    lambda parts: parts[1] if len(parts) > 1 else pd.NA
)

csv[["Make_n_model", "Make", "Model"]].head()
```

## Проверка новых столбцов

```python
csv = csv.drop(columns=["Make_n_model"])
```

## Сохранение JSON

```python
csv.to_json(
    "../data/auto.json",
    orient="records",
    indent=2
)
```
