## Шаг 2

```python
df = pd.read_csv(
    "../data/fines.csv"
)
df
```

## Определение функции обработки строк

```python
def calculate_with_loop(df):
    result = []

    for i in range(0, len(df)):
        row = df.iloc[i]

        value = row["Fines"] / row["Refund"] * row["Year"]
        result.append(value)

    return result
```

## Измерение времени вызова

```python
%%timeit -r 3 -n 1
df["loop_result"] = calculate_with_loop(df)
```

## Определение функции обработки строк

```python
def calculate_with_iterrows(df):
    result = []

    for _, row in df.iterrows():
        value = row["Fines"] / row["Refund"] * row["Year"]
        result.append(value)

    return result
```

## Измерение времени вызова

```python
%%timeit -r 3 -n 1
df["iterrows_result"] = calculate_with_iterrows(df)
```

## Шаг 3

```python
%%timeit -r 3 -n 1 
df["apply_result"] = df.apply(
    lambda row: row["Fines"] / row["Refund"] * row["Year"],
    axis=1
)
```

## Шаг 4

```python
%%timeit -r 3 -n 1 
df["series_result"] = (
    df["Fines"] / df["Refund"] * df["Year"]
)
```

## Шаг 5

```python
%%timeit -r 3 -n 1 
df["values_result"] = (
    df["Fines"].values
    / df["Refund"].values
    * df["Year"].values
)
```
