```python
import pandas as pd
import gc
```

```python
df = pd.read_csv(
    "../ex04/fines.csv"
)
df
```

```python
%%timeit 
def calculate_with_loop(df):
    result = []

    for i in range(0, len(df)):
        row = df.iloc[i]

        value = row["Fines"] / row["Refund"] * row["Year"]
        result.append(value)

    return result
df["loop_result"] = calculate_with_loop(df)
```

```python
%%timeit 
def calculate_with_iterrows(df):
    result = []

    for _, row in df.iterrows():
        value = row["Fines"] / row["Refund"] * row["Year"]
        result.append(value)

    return result
df["iterrows_result"] = calculate_with_iterrows(df)
```

```python
%%timeit 
df["apply_result"] = df.apply(
    lambda row: row["Fines"] / row["Refund"] * row["Year"],
    axis=1
)
```

```python
%%timeit 
df["series_result"] = (
    df["Fines"] / df["Refund"] * df["Year"]
)
```

```python
%%timeit 
df["values_result"] = (
    df["Fines"].values
    / df["Refund"].values
    * df["Year"].values
)
```

```python
optimized_df = df.copy()
```

```python
optimized_df.dtypes
```

```python
float_columns = optimized_df.select_dtypes(
    include=["float64"]
).columns

float_columns
```

```python
optimized_df[float_columns] = (
    optimized_df[float_columns].astype("float32")
)
```

```python
optimized_df.dtypes
```

```python
int_columns = optimized_df.select_dtypes(
    include=["int64"]
).columns

int_columns
```

```python
for column in int_columns:
    optimized_df[column] = pd.to_numeric(
        optimized_df[column],
        downcast="integer"
    )
```

```python
optimized_df.info(memory_usage='deep')
```

```python
text_columns = optimized_df.select_dtypes(
    include=["object", "string"]
).columns

text_columns
```

```python
for column in text_columns:
    optimized_df[column] = optimized_df[column].astype("category")
optimized_df.dtypes
```
