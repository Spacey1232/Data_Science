## Шаг 1

```python
import pandas as pd
import numpy as np
import requests
```

## Шаг 2

```python
fines = pd.read_json(
    "../data/auto.json"
)
fines.head()
```

## Шаг 3

```python
pd.options.display.float_format = "{:.2f}".format
fines.head()
```

## Шаг 4

```python
sample = fines.sample(
    n=200,
    random_state=21
)

sample.head()
sample.shape
```

## Шаг 5

```python
concat_rows = pd.concat(
    [fines, sample],
    ignore_index=True
)

concat_rows.head()
```

## Шаг 6

```python
np.random.seed(21)
years = pd.Series(
    np.random.randint(
        low=1980,
        high=2020,
        size=len(concat_rows)
    ),
    name="Year"
)

years.head()
```

## Шаг 7

```python
fines = pd.concat(
    [concat_rows, years],
    axis=1
)

fines.head()
```

## Шаг 8

```python
fines.shape
fines["Year"].min()
fines["Year"].max()
```

## Шаг 9

```python
surname_raw = pd.read_json(
    "../data/surname.json"
)

surname_raw.head()
```

## Шаг 10

```python
surname_table = pd.DataFrame(
    data=surname_raw.values[1:],
    columns=surname_raw.values[0]
)

surname_table.head()
```

## Шаг 11

```python
car_numbers = pd.Series(
    sample["CarNumber"].unique(),
    name="CarNumber"
)

car_numbers.head()
```

## Шаг 12

```python
surname_series = surname_table["NAME"].sample(
    n=len(car_numbers),
    replace=True,
    random_state=21,
    ignore_index=True
)
surname_series
```

## Шаг 13

```python
surname_series = surname_series.str.replace(
    r"[^A-Za-z]",
    "",
    regex=True
)

surname_series.name = "SURNAME"
```

## Шаг 14

```python
owners = pd.concat(
    [car_numbers, surname_series],
    axis=1
)

owners.head()
```

## Шаг 15

```python
owners.shape
owners.isna().sum()
```

## Шаг 16

```python
new_fines = pd.DataFrame(
    {
        "CarNumber": [
            "A001AA777RUS",
            "B002BB777RUS",
            "C003CC777RUS",
            "D004DD777RUS",
            "E005EE777RUS"
        ],
        "Refund": [
            1.0,
            2.0,
            1.0,
            2.0,
            1.0
        ],
        "Fines": [
            1500.0,
            4200.0,
            8300.0,
            2100.0,
            5600.0
        ],
        "Make": [
            "Ford",
            "Toyota",
            "BMW",
            "Volkswagen",
            "Honda"
        ],
        "Model": [
            "Focus",
            "Camry",
            "X5",
            "Golf",
            "CRV"
        ],
        "Year": [
            2015,
            2018,
            2019,
            2012,
            2017
        ]
    }
)

new_fines
```

## Шаг 17

```python
fines = pd.concat(
    [fines, new_fines],
    ignore_index=True
)
fines.tail(5)
```

## Шаг 18

```python
owners = owners.iloc[:-20]
new_owners = pd.DataFrame(
    {
        "CarNumber": [
            "Z101ZZ777RUS",
            "Y202YY777RUS",
            "W303WW777RUS"
        ],
        "SURNAME": [
            "SMITH",
            "WILLIAMS",
            "TAYLOR"
        ]
    }
)
owners = pd.concat(
    [owners, new_owners],
    ignore_index=True
)
```

## Шаг 19

```python
fines_owners_inner = fines.merge(
    owners,
    on="CarNumber",
    how="inner"
)

fines_owners_inner.head()
```

## Шаг 20

```python
fines_owners_outer = fines.merge(
    owners,
    on="CarNumber",
    how="outer"
)

fines_owners_outer.head()
```

## Шаг 21

```python
fines_owners_left = fines.merge(
    owners,
    on="CarNumber",
    how="left"
)

fines_owners_left.head()
```

## Шаг 22

```python
fines_owners_right = fines.merge(
    owners,
    on="CarNumber",
    how="right"
)

fines_owners_right.head()
```

## Шаг 23

```python
print("inner:", fines_owners_inner.shape)
print("outer:", fines_owners_outer.shape)
print("left:", fines_owners_left.shape)
print("right:", fines_owners_right.shape)
```

## Шаг 24

```python
fines_pivot = fines.pivot_table(
    index=["Make", "Model"],
    columns="Year",
    values="Fines",
    aggfunc="sum"
)

fines_pivot
```
