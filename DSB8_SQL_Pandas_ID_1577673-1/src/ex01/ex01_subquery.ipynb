## Импорт библиотек

```python
import pandas as pd
import sqlite3
```

## Подключение к базе

```python
connection = sqlite3.connect("../data/checking-logs.sqlite")
```

## Структура таблицы

```python
pd.io.sql.read_sql(
    "PRAGMA table_info(checker);",
    connection
)
```

## Обработка и анализ данных

```python
pd.io.sql.read_sql(
    "SELECT * FROM checker LIMIT 10;",
    connection
)
```

## Обработка и анализ данных

```python
result = pd.io.sql.read_sql(
    """
    SELECT COUNT(*) AS cnt
    FROM pageviews
    WHERE uid IN (
        SELECT uid
        FROM checker
        WHERE status = 'ready'
          AND numTrials = 1
          AND labname IN (
              'laba04', 'laba04s', 'laba05',
              'laba06', 'laba06s', 'project1'
          )
    );
    """,
    connection
)

result
```
