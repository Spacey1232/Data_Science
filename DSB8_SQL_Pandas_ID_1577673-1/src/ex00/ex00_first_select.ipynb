## Импорт библиотек

```python
import pandas as pd
import sqlite3
```

## Подключение к базе

```python
connection = sqlite3.connect("../data/checking-logs.sqlite")
```

## Структура таблицы

```python
pd.io.sql.read_sql(
    "PRAGMA table_info(pageviews);",
    connection
)
```

## Обработка и анализ данных

```python
pd.io.sql.read_sql(
    "SELECT * FROM pageviews LIMIT 10;",
    connection
)
```

## Обработка и анализ данных

```python
pageviews = pd.io.sql.read_sql(
    """
    SELECT uid, datetime
    FROM pageviews
    WHERE uid GLOB 'user_*'
    ORDER BY uid ASC;
    """,
    connection,
    index_col="datetime",
    parse_dates=["datetime"]
)

pageviews.head()
```
