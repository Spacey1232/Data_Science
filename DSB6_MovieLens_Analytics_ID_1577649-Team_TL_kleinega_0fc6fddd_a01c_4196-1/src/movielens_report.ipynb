# Анализ данных MovieLens

Цель отчёта — исследовать оценки пользователей, фильмы, жанры, теги и дополнительные данные IMDb.

Для анализа используются первые 1000 записей каждого CSV-файла. Все вычисления выполняются классами и методами из модуля `movielens_analysis.py`.

```python
from movielens_analysis import Ratings, Tags, Movies, Links, average, median

RATINGS_PATH = "../datasets/ml-latest-small/ratings.csv"
TAGS_PATH = "../datasets/ml-latest-small/tags.csv"
MOVIES_PATH = "../datasets/ml-latest-small/movies.csv"
LINKS_PATH = "../datasets/ml-latest-small/links.csv"

%timeit -n 1 -r 1 Ratings(RATINGS_PATH)
ratings = Ratings(RATINGS_PATH)

%timeit -n 1 -r 1 Tags(TAGS_PATH)
tags = Tags(TAGS_PATH)

%timeit -n 1 -r 1 Movies(MOVIES_PATH)
movies = Movies(MOVIES_PATH)

%timeit -n 1 -r 1 Links(LINKS_PATH)
links = Links(LINKS_PATH)
```

## 1. Анализ оценок фильмов

Сначала посмотрим, в какие годы пользователи ставили оценки и как распределились сами оценки. Затем найдём самые часто оцениваемые фильмы, лидеров по среднему и медианному рейтингу, а также фильмы с наиболее противоречивыми оценками.

```python
%timeit -n 1 -r 3 ratings.movies.dist_by_year()
%timeit -n 1 -r 3 ratings.movies.dist_by_rating()
%timeit -n 1 -r 3 ratings.movies.top_by_num_of_ratings(10)
%timeit -n 1 -r 3 ratings.movies.top_by_ratings(10, average)
%timeit -n 1 -r 3 ratings.movies.top_by_ratings(10, median)
%timeit -n 1 -r 3 ratings.movies.top_controversial(10)

{
    "Распределение по годам": ratings.movies.dist_by_year(),
    "Распределение оценок": ratings.movies.dist_by_rating(),
    "Больше всего оценок": ratings.movies.top_by_num_of_ratings(10),
    "Топ по среднему": ratings.movies.top_by_ratings(10, average),
    "Топ по медиане": ratings.movies.top_by_ratings(10, median),
    "Самые спорные": ratings.movies.top_controversial(10),
}
```

### Вывод по оценкам фильмов

Высокий средний рейтинг ещё не означает настоящую популярность: фильм мог получить только одну высокую оценку. Поэтому средний рейтинг необходимо рассматривать вместе с количеством оценок.

Метод `top_controversial()` использует дисперсию. Чем она выше, тем сильнее мнения пользователей о фильме отличаются друг от друга.

## 2. Анализ пользователей

Теперь сравним пользователей по количеству поставленных оценок, среднему и медианному значению оценок, а также по степени противоречивости их мнений.

```python
%timeit -n 1 -r 3 ratings.users.dist_by_num_of_ratings()
%timeit -n 1 -r 3 ratings.users.dist_by_ratings(average)
%timeit -n 1 -r 3 ratings.users.dist_by_ratings(median)
%timeit -n 1 -r 3 ratings.users.top_controversial(10)

{
    "Количество оценок": ratings.users.dist_by_num_of_ratings(),
    "Средняя оценка": ratings.users.dist_by_ratings(average),
    "Медианная оценка": ratings.users.dist_by_ratings(median),
    "Самые противоречивые": ratings.users.top_controversial(10),
}
```

### Вывод по пользователям

Пользователи отличаются не только активностью, но и стилем оценивания. Большая дисперсия означает, что пользователь применяет широкий диапазон оценок, а небольшая — что он оценивает фильмы более однообразно.

## 3. Анализ тегов

Теги — это свободные текстовые описания фильмов. Исследуем теги с наибольшим количеством слов, самые длинные и популярные теги, а также выполним поиск по заданному слову.

```python
%timeit -n 1 -r 3 tags.most_words(10)
%timeit -n 1 -r 3 tags.longest(10)
%timeit -n 1 -r 3 tags.most_words_and_longest(10)
%timeit -n 1 -r 3 tags.most_popular(10)
%timeit -n 1 -r 3 tags.tags_with("love")

{
    "Больше всего слов": tags.most_words(10),
    "Самые длинные": tags.longest(10),
    "Пересечение списков": tags.most_words_and_longest(10),
    "Самые популярные": tags.most_popular(10),
    "Содержат love": tags.tags_with("love"),
}
```

### Вывод по тегам

Количество слов и длина строки — разные характеристики. Длинный тег не обязательно содержит больше всего слов. Популярные теги показывают повторяющиеся пользовательские описания, а поиск позволяет изучать конкретные темы и эмоции.

## 4. Анализ фильмов и жанров

Изучим распределение фильмов по годам выпуска и жанрам, а затем найдём фильмы, которые одновременно относятся к наибольшему количеству жанров.

```python
%timeit -n 1 -r 3 movies.dist_by_release()
%timeit -n 1 -r 3 movies.dist_by_genres()
%timeit -n 1 -r 3 movies.most_genres(10)

{
    "Фильмы по годам": movies.dist_by_release(),
    "Фильмы по жанрам": movies.dist_by_genres(),
    "Больше всего жанров": movies.most_genres(10),
}
```

### Вывод по фильмам

Распределение по годам показывает, фильмы каких периодов преобладают в учебной выборке. Один фильм может относиться сразу к нескольким жанрам, поэтому сумма значений в распределении по жанрам превышает общее количество фильмов.

## 5. Связь с IMDb

Файл `links.csv` связывает внутренний `movieId` с идентификаторами IMDb и TMDb. Для нескольких фильмов попробуем получить режиссёра, бюджет, мировые сборы и продолжительность.

IMDb может ограничивать автоматические запросы. Поэтому отсутствие отдельных значений не обязательно означает ошибку в данных MovieLens.

```python
IMDB_FIELDS = [
    "Director",
    "Budget",
    "Cumulative Worldwide Gross",
    "Runtime",
]

%timeit -n 1 -r 1 links.get_imdb([1, 2, 3], IMDB_FIELDS)
links.imdb_info
```

## 6. Анализ информации IMDb

На основе полученных данных определим наиболее часто встречающихся режиссёров, самые дорогие и прибыльные фильмы, наиболее длинные фильмы и фильмы с максимальной стоимостью одной минуты.

```python
%timeit -n 1 -r 3 links.top_directors(10)
%timeit -n 1 -r 3 links.most_expensive(10)
%timeit -n 1 -r 3 links.most_profitable(10)
%timeit -n 1 -r 3 links.longest(10)
%timeit -n 1 -r 3 links.top_cost_per_minute(10)

{
    "Режиссёры": links.top_directors(10),
    "Самые дорогие": links.most_expensive(10),
    "Самые прибыльные": links.most_profitable(10),
    "Самые длинные": links.longest(10),
    "Стоимость минуты": links.top_cost_per_minute(10),
}
```

### Вывод по IMDb

Большой бюджет сам по себе не гарантирует финансовый успех. Поэтому прибыль рассчитывается отдельно как разница между мировыми сборами и бюджетом. Стоимость минуты показывает отношение бюджета к продолжительности фильма.

Если итоговые словари пустые, наиболее вероятная причина — IMDb не предоставил данные автоматическому запросу.

# Итоговый вывод

Набор MovieLens позволяет исследовать несколько сторон кинематографа:

- активность и предпочтения пользователей;
- популярность и спорность фильмов;
- распределение фильмов по годам и жанрам;
- содержание пользовательских тегов;
- связь MovieLens с внешними источниками через IMDb ID.

Кроме того, высокий средний рейтинг не всегда означает настоящую популярность: необходимо одновременно учитывать количество оценок и разброс пользовательских мнений.
